## 2. ชนิดข้อมูล 4 ชนิดที่ต้องรู้จัก

| ชนิด | ตัวอย่าง | ใช้ทำอะไร |
|---|---|---|
| `int` | `42` | จำนวนนับ (จำนวนแถว, จำนวนครั้ง) |
| `float` | `42.5` | ตัวเลขมีทศนิยม (ราคา, ค่าการวัด) |
| `str` | `"ข้าวผัดกะเพรา"` | ข้อความ — ต้องมีเครื่องหมายคำพูด |
| `bool` | `True` / `False` | คำตอบ ใช่/ไม่ — หัวใจของการกรองข้อมูล |

```python
count = 3              # int
price = 45.5           # float
dish = "ข้าวผัดกะเพรา"   # str
spicy = True           # bool

print(type(count), type(price))
print(type(dish), type(spicy))
```

```python
# ชนิดไม่ปนกัน: ข้อความบวกตัวเลขตรง ๆ พัง
# ("50" กับ 50 คนละเรื่อง — อันแรกคือข้อความ ไม่ใช่จำนวน)
try:
    print("50" + 50)
except TypeError as err:
    print("พังตามคาด:", err)
```

## 3. f-string — ฝังค่าลงในข้อความ

ข้อมูลที่ดีต้อง "สื่อสาร" ได้ f-string ใส่ค่าตัวแปรลงในข้อความด้วย `{}` และ
จัดรูปตัวเลขให้สวยได้ในตัว (`:,` ใส่คอมมา · `.1f` ทศนิยม 1 ตำแหน่ง)

```python
name = "นักวิเคราะห์"
price = 45.5
print(f"{name}กินข้าว {price} บาท")
print(f"ราคาจัดรูป: {price:.1f} บาท · จำนวนใหญ่: {1234567:,}")
```

```python
# f-string คำนวณในปีกได้ด้วย — ไม่ต้องสร้างตัวแปรกั้นเสมอ
print(f"คนละครึ่ง: {price / 2} บาท")
```

## 4. list — รายการเรียงลำดับ

ข้อมูล "รอบตัวเรา" มักเป็นรายการ: รายจ่ายแต่ละมื้อ อุณหภูมิแต่ละวัน
Python เก็บรายการด้วย `list` — และการนับตำแหน่ง **เริ่มที่ 0** เสมอ

```python
lunch = [45, 60, 50, 120, 40]   # รายจ่ายมื้อกลางวัน จันทร์–ศุกร์

print(len(lunch))       # มีกี่รายการ
print(lunch[0])         # รายการแรก (นับจาก 0!)
print(lunch[-1])        # รายการสุดท้าย
print(lunch[1:4])       # ตำแหน่ง 1 ถึง 3
```

```python
# เพิ่มรายการใหม่ท้าย list ด้วย .append()
lunch.append(65)        # วันเสาร์กินข้าวร้านดัง
print(lunch)
print(sum(lunch), max(lunch), min(lunch))   # รวม · มากสุด · น้อยสุด
```

## 5. dict — เก็บแบบมีป้ายชื่อ

`list` เก็บตามลำดับ แต่หลายข้อมูลมาพร้อมชื่อกำกับ: ราคาแต่ละเมนู คะแนนแต่ละคน
`dict` จับคู่ **key → value** เข้าถึงด้วยชื่อ ไม่ต้องจำตำแหน่ง

```python
menu = {"ข้าวผัดกะเพรา": 50, "ข้าวมันไก่": 60, "ก๋วยเตี๋ยว": 45}

print(menu["ข้าวมันไก่"])    # ดึงค่าด้วย key
menu["สลัด"] = 65           # เพิ่มคู่ key-value ใหม่
print(menu)
```

```python
# ถ้าถาม key ที่ไม่มี — พัง! เช็คก่อนด้วย in
print("พิซซ่า" in menu)
```

## 6. for — ให้คอมพิวเตอร์ทำซ้ำแทนเรา

จดรายจ่ายทุกวันทีละบรรทัดด้วยมือ 5 วันพอไหว แต่ 5,000 แถวล่ะ —
นี่คือเหตุผลที่เราเขียนโค้ด: บอกสูตรครั้งเดียว ให้มันวนเอง

```python
total = 0
for cost in lunch:          # วนทีละค่าใน list
    total = total + cost    # สะสมผลรวม
    print(cost, "->", total)
```

```python
# วนสะสมข้างบนย่อได้: sum() ทำงานเดียวกันในบรรทัดเดียว
print("รวมทั้งสัปดาห์:", sum(lunch))
print("เฉลี่ยต่อวัน:", round(sum(lunch) / len(lunch), 1))
```

## 7. if และฟังก์ชันของเราเอง

`if` คือการตัดสินใจ: เงื่อนไขจริง → ทำ ไม่จริง → ข้าม ส่วน `def` คือการ
ห่อคำสั่งชุดหนึ่งให้เป็น "เครื่องมือมีชื่อ" เรียกใช้ซ้ำได้ทุกเมื่อ

```python
budget = 300
if sum(lunch) > budget:
    print("เกินงบสัปดาห์นี้!")
else:
    print("งบยังอยู่")
```

```python
def status(total, budget):
    """บอกสถานะงบรายสัปดาห์"""
    if total > budget:
        return f"เกินงบ {total - budget} บาท"
    return f"ยังเหลือ {budget - total} บาท"


print(status(sum(lunch), 300))
print(status(sum(lunch), 400))
```

```python
# def ห่อสูตรให้เป็นเครื่องมือ — แก้กฎที่เดียว ใช้ได้ทุกที่
# (โมดูลถัดไปจะเจอ groupby: คิดสูตรครั้งเดียว ใช้กับทุกกลุ่ม เช่นกัน)
```

## 8. ข้อมูลจริงอยู่ไหน — pathlib.Path

ทุกอย่างที่ผ่านมาเป็นข้อมูลที่เราพิมพ์เอง ของจริงอยู่ใน **ไฟล์** `Path`
คือ "ที่อยู่ของไฟล์" ที่ใช้ได้ทั้ง Windows และ macOS — โมดูลถัดไปจะใช้มัน
ชี้หาไฟล์ข้อมูลจริงใน repo นี้

```python
from pathlib import Path  # noqa: E402 — ที่อยู่ไฟล์ ใช้ในหัวข้อนี้


def repo_root() -> Path:
    """หา repo root โดยไล่ขึ้นจาก cwd."""
    for p in (Path.cwd(), *Path.cwd().parents):
        if (p / "requirements-offline.txt").exists():
            return p
    raise SystemExit(
        "ไม่พบ repo root — เปิด JupyterLab จาก repo root (uv run jupyter lab)")


course = repo_root()
print("repo อยู่ที่:", course)
```

```python
# ข้อมูลจริงของคอร์สอยู่ที่ datasets/ — เปิดดูว่ามีชุดไหนรออยู่
for p in sorted((course / "datasets").iterdir()):
    if p.is_dir():
        print(p.name)
```
