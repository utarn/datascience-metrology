## ส่วนที่ 1 — ตัวแปรและชนิดข้อมูล

```python
dish = "ข้าวราดแกง"     # str
price = 55.0           # float
spicy = True           # bool

print(type(dish), type(price), type(spicy))
```

**1.1** `"45.50"` คือ `str` (ข้อความ 5 ตัวอักษร) ส่วน `45.50` คือ `float`
(จำนวนจริง) ถ้าไฟล์จริงเก็บราคาเป็นข้อความ การบวก/เฉลี่ยจะพังหรือผิด —
ต้องแปลงเป็นตัวเลขก่อน (วิธีแปลงจริงคือ `.astype()` ที่จะได้เรียนใน M3)

## ส่วนที่ 2 — list

```python
coffee = [45, 50, 45, 60, 0, 55, 45]
coffee.append(50)

print("จำนวนรายการ:", len(coffee))
print("ราคารวม:", sum(coffee))
```

**2.1** รู้ไม่ได้ — `list` จำแค่ลำดับ ไม่จำความหมายของแต่ละค่า ขาด
"ป้ายชื่อ" บอกว่าค่าไหนเป็นของวันไหน โครงสร้างที่แก้ได้คือ `dict`
(key = วัน, value = ราคา) หรือตารางที่มีคอลัมน์วันที่ — ซึ่งคือรูปแบบ
ของข้อมูลจริงในโมดูลถัดไป

## ส่วนที่ 3 — dict

```python
menu = {"ข้าวผัดกะเพรา": 50, "ข้าวมันไก่": 60, "ก๋วยเตี๋ยว": 45}

chicken_rice = menu["ข้าวมันไก่"]
print(f"ข้าวมันไก่ราคา {chicken_rice} บาท")
```

**3.1** `list` เก็บตามลำดับ เข้าถึงด้วยตำแหน่ง เหมาะกับข้อมูลลำดับเดียวกัน
(รายจ่ายรายวัน) `dict` เก็บแบบมีชื่อกำกับ เข้าถึงด้วย key เหมาะกับข้อมูลที่
ต้องหยิบด้วยชื่อ (ราคาต่อเมนู) — ตัวอย่างจากงาน: รหัสพัสดุ → ชื่อพัสดุ เป็น dict
ส่วนประวัติรับของรายเดือนเป็น list

## ส่วนที่ 4 — for และฟังก์ชันของตัวเอง

```python
expenses = [120, 85, 60, 45, 200, 95, 110]

total = 0
for cost in expenses:
    total = total + cost
print("สะสมเอง:", total, "· sum():", sum(expenses))
```

```python
def daily_report(total):
    """สรุปรายจ่ายของหนึ่งวันเป็นข้อความ"""
    if total == 0:
        return "วันนี้ไม่ได้ใช้จ่าย"
    return f"วันนี้ใช้ {total} บาท"


for cost in expenses:
    print(daily_report(cost))
```
